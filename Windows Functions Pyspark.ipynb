```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T

spark = SparkSession.builder.getOrCreate()

data = [
    (1, 101, "2024-01-01 10:00:00", "Electronics", 1200, "Mumbai"),
    (2, 101, "2024-01-02 15:00:00", "Grocery", 250, "Mumbai"),
    (3, 102, "2024-01-01 09:15:00", "Fashion", 800, "Delhi"),
    (4, 102, "2024-01-05 17:30:00", "Electronics", 1500, "Delhi"),
    (5, 103, "2024-01-02 11:00:00", "Grocery", 300, "Bangalore"),
    (6, 103, "2024-01-02 11:45:00", "Grocery", 150, "Bangalore"),
    (7, 101, "2024-01-05 18:10:00", "Fashion", 2000, "Mumbai"),
    (8, 104, "2024-01-03 14:10:00", "Fashion", 450, "Chennai"),
    (9, 104, "2024-01-03 16:10:00", "Electronics", 1000, "Chennai"),
    (10, 101, "2024-01-08 09:00:00", "Grocery", 350, "Mumbai"),
    (11, 102, "2024-01-09 19:30:00", "Fashion", 700, "Delhi"),
    (12, 103, "2024-01-05 18:00:00", "Electronics", 2000, "Bangalore"),
]

schema = T.StructType([
    T.StructField("transaction_id", T.IntegerType(), False),
    T.StructField("customer_id", T.IntegerType(), False),
    T.StructField("transaction_date", T.StringType(), False),
    T.StructField("product_category", T.StringType(), False),
    T.StructField("amount", T.DoubleType(), False),
    T.StructField("city", T.StringType(), False)
])

df = spark.createDataFrame(data, schema)
df.show(truncate=False)
```
